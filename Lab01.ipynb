```python
MODEL = "gemini-3.5-flash-lite"

def count_tokens(text, model=MODEL):
    return client.models.count_tokens(
        model=model,
        contents=text
    ).total_tokens


texts = {
    "EN": EN_TEXT,
    "RU": RU_TEXT,
    "KY": KY_TEXT,
    "Код": CODE_TEXT,
    "Числа": NUMBERS_TEXT,
}

rows = []

for name, text in texts.items():
    tokens = count_tokens(text)

    rows.append({
        "Текст": name,
        "Символов": len(text),
        "Слов": len(text.split()),
        "Токенов": tokens,
        "Симв./токен": round(len(text) / tokens, 2),
        "Токенов/слово": round(
            tokens / len(text.split()), 2
        ),
    })

    time.sleep(1)

df_tokens = pd.DataFrame(rows)
display(df_tokens)
```

```python
import time

prompt = (
    "Explain in simple English how an API works. "
    "Give three short examples."
)

start_time = time.perf_counter()

response_stream = client.models.generate_content_stream(
    model=MODEL,
    contents=prompt,
)

first_chunk_time = None
chunks = []
output_text = ""

for chunk in response_stream:
    if first_chunk_time is None:
        first_chunk_time = time.perf_counter()

    chunks.append(chunk)

    if chunk.text:
        output_text += chunk.text

end_time = time.perf_counter()

ttft = first_chunk_time - start_time
total_latency = end_time - start_time

print("Ответ модели:")
print(output_text)
print()
print(f"TTFT: {ttft:.3f} сек.")
print(f"Общее время: {total_latency:.3f} сек.")
```

```python
# Подсчитываем токены для текста, который модель сгенерировала
token_info = client.models.count_tokens(
    model=MODEL,
    contents=output_text
)

output_tokens = token_info.total_tokens

# Оцениваем скорость после получения первого фрагмента
generation_time = total_latency - ttft

if generation_time > 0:
    tokens_per_second = output_tokens / generation_time
else:
    tokens_per_second = 0

print("Выходных токенов (оценка):", output_tokens)
print(f"Время после первого фрагмента: {generation_time:.3f} сек.")
print(f"Скорость генерации (оценка): {tokens_per_second:.2f} токенов/сек.")
```

```python
start_time = time.perf_counter()

response = client.models.generate_content(
    model=MODEL,
    contents=prompt,
)

end_time = time.perf_counter()

usage = response.usage_metadata

print("Время обычного запроса:", round(end_time - start_time, 3), "сек.")
print("Входные токены:", usage.prompt_token_count)
print("Выходные токены:", usage.candidates_token_count)
print("Всего токенов:", usage.total_token_count)
```

```python
prompt_temp = (
    "Give one creative name for a small coffee shop. "
    "Answer with the name only."
)

for temp in [0.0, 0.7, 1.3]:
    response = client.models.generate_content(
        model=MODEL,
        contents=prompt_temp,
        config={
            "temperature": temp,
            "max_output_tokens": 50,
        },
    )

    print(f"Temperature = {temp}")
    print(response.text)
    print("-" * 40)
    time.sleep(1)
```

```python
repeat_prompt = (
    "Explain in one short sentence what an API is."
)

for i in range(3):
    response = client.models.generate_content(
        model=MODEL,
        contents=repeat_prompt,
        config={
            "temperature": 0.0,
            "max_output_tokens": 60,
        },
    )

    print(f"Run {i + 1}:")
    print(response.text)
    print("-" * 40)
    time.sleep(1)
```
